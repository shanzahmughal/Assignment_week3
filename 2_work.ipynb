```python
import pandas as pd

df = pd.read_csv("students (1) (1).csv")

print("Mean Marks:", df["Marks"].mean())
print("Median Marks:", df["Marks"].median())
```

```python
print("Age with outlier:", df['Age'].mean())
cleaned_age = df[df['Age'] < 200]['Age']
print("Age without outlier:", cleaned_age.mean(), cleaned_age.median())
```
